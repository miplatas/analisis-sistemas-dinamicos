# 1.1 Señales y sistemas

Una **señal** es una representación física que cambia con respectoa a alguna variable dependiente. 

Un **sistema** es un conjunto de elementos que transforma una señal de entrada en una de salida. Una señal continua se define para todo tiempo $t\in \Re$, aunque aquí (en la PC) la representamos mediante muestras.

El escalón unitario $u(t)$ vale cero para $t<0$ y uno para $t\geq 0$. La señal $r(t)=t u(t)$ es una rampa. En un circuito, por ejemplo, $u(t)$ puede modelar la aplicación de un voltaje.

**Objetivo:** identificar el papel de la variable independiente y reconocer señales elementales.

```python
%matplotlib inline
import numpy as np
from sistemas_dinamicos import escalon, graficar

t = np.linspace(-2, 4, 601)
u = escalon(t)
r = t * u
graficar(t, {"Escalón": u, "Rampa": r}, titulo="Señales elementales");
```

Antes de $t=0$ ambas señales son cero (señales causales). El escalón cambia abruptamente y la rampa crece linealmente.

**Practica:** desplaza el escalón para que inicie en $t=1$ (atraso) con `escalon(t, inicio=1)` y construye una rampa que empiece en ese instante.
