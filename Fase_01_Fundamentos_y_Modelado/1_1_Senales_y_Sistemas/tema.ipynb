# 1.1 Señales y sistemas

Una **señal** asigna un valor a cada instante; un **sistema** transforma una entrada en una salida. Una señal continua se define para todo tiempo, aunque aquí la representamos mediante muestras.

El escalón unitario $u(t)$ vale cero para $t<0$ y uno para $t\geq 0$. La señal $r(t)=t u(t)$ es una rampa. En un circuito, por ejemplo, $u(t)$ puede modelar la aplicación de un voltaje.

**Objetivo:** identificar el papel de la variable independiente y reconocer señales elementales.

```python
%matplotlib inline
import numpy as np
from sistemas_dinamicos import escalon, graficar

t = np.linspace(-2, 4, 601)
u = escalon(t)
r = t * u
graficar(t, {"Escalón": u, "Rampa": r}, titulo="Señales elementales");
```

Antes de $t=0$ ambas señales son cero. El escalón cambia abruptamente y la rampa crece linealmente. La separación temporal entre muestras afecta la representación, no la definición continua.

**Practica:** desplaza el escalón a $t=1$ con `escalon(t, inicio=1)` y construye una rampa que empiece en ese instante.
