# 1.4 Función de transferencia y espacio de estados

Para un sistema LTI con condiciones iniciales nulas, $G(s)=Y(s)/U(s)$. Un modelo de estados se escribe $\dot x=Ax+Bu$, $y=Cx+Du$ y produce $G(s)=C(sI-A)^{-1}B+D$.

El sistema $\dot x=-2x+u$, $y=x$ tiene $G(s)=1/(s+2)$. Las representaciones describen la misma relación entrada-salida; el estado también permite especificar condiciones iniciales.

**Objetivo:** relacionar ambos modelos.

```python
%matplotlib inline
import numpy as np
from scipy import signal
from sistemas_dinamicos import graficar

A, B, C, D = [[-2]], [[1]], [[1]], [[0]]
modelo_estado = signal.StateSpace(A, B, C, D)
modelo_transferencia = signal.TransferFunction([1], [1, 2])
t = np.linspace(0, 5, 501)
_, y_estado = signal.step(modelo_estado, T=t)
_, y_transferencia = signal.step(modelo_transferencia, T=t)
graficar(t, {"Estados": y_estado, "Transferencia": y_transferencia}, titulo="Dos modelos equivalentes");
print("Diferencia máxima:", np.max(np.abs(y_estado - y_transferencia)))
```

Ambas respuestas coinciden para estado inicial nulo y tienden a $1/2$. La equivalencia mostrada es entrada-salida.

**Practica:** cambia $A$ a $-1$ y determina el denominador de la función de transferencia correspondiente.
