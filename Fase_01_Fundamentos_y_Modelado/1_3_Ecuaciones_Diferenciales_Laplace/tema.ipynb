# 1.3 Ecuaciones diferenciales y transformada de Laplace

Considera $\dot y+y=u(t)$ con $y(0)=0$ y entrada escalón. La transformada unilateral cumple $\mathcal L\{\dot y\}=sY(s)-y(0)$. Por tanto $(s+1)Y(s)=1/s$ y

$$Y(s)=\frac{1}{s(s+1)},\qquad y(t)=1-e^{-t},\quad t\geq0.$$

Las condiciones iniciales importan: la función de transferencia por sí sola corresponde a condiciones iniciales nulas.

**Objetivo:** pasar del modelo diferencial a una solución temporal.

```python
%matplotlib inline
import numpy as np
import sympy as sp
from sistemas_dinamicos import graficar

s = sp.symbols("s")
t_simbolico = sp.symbols("t", positive=True)
Y = 1 / (s * (s + 1))
y = sp.inverse_laplace_transform(Y, s, t_simbolico)
print("Y(s) =", Y, "; y(t) =", y)
t = np.linspace(0, 5, 501)
graficar(t, {"y(t)": sp.lambdify(t_simbolico, y, "numpy")(t)}, titulo="Respuesta al escalón");
```

La salida empieza en cero y tiende a uno; la constante de tiempo es $1$ s. El resultado se interpreta para $t\geq0$.
