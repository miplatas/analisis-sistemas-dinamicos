# 3.3 Error, paso y estabilidad numérica

Para $\dot x=\lambda x$, Euler produce $x_{k+1}=(1+h\lambda)x_k$. Con $\lambda<0$, la solución exacta decae, pero Euler solo reproduce ese decaimiento si $|1+h\lambda|<1$.

Con $\lambda=-1$, el intervalo estable es $0<h<2$. Un paso demasiado grande puede generar crecimiento numérico en un sistema continuo estable.

**Objetivo:** distinguir estabilidad del sistema y del método numérico.

```python
%matplotlib inline
import numpy as np
from sistemas_dinamicos import graficar, integrar

t = np.arange(0, 22, 2.5)
aproximada = integrar(lambda t, x: -x, t, [1], metodo="euler")[:, 0]
graficar(t, {"Euler, h=2.5": aproximada, "Exacta": np.exp(-t)}, titulo="Inestabilidad numérica");
print("Factor de amplificación:", 1 - 2.5)
```

El factor $-1.5$ alterna de signo y crece en magnitud, aunque $e^{-t}$ decae. La condición calculada corresponde a Euler y a este problema de prueba, no a cualquier integrador.

**Practica:** prueba $h=1$, $h=2$ y $h=2.5$; distingue decaimiento, caso límite e inestabilidad.
