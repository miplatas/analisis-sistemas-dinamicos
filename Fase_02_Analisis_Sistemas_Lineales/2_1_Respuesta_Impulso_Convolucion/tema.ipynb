# 2.1 Respuesta al impulso y convolución

En un sistema LTI causal, la salida con estado inicial nulo es $y(t)=\int_0^t h(\tau)u(t-\tau)\,d\tau$, donde $h$ es la respuesta al impulso. Para $G(s)=1/(s+1)$, $h(t)=e^{-t}$ para $t\geq0$.

Con entrada escalón, $y(t)=1-e^{-t}$. La suma numérica aproxima la integral con paso $\Delta t$.

**Objetivo:** conectar respuesta al impulso y respuesta a una entrada.

```python
%matplotlib inline
import numpy as np
from sistemas_dinamicos import convolucion, escalon, graficar

t = np.linspace(0, 5, 1001)
h = np.exp(-t)
y = convolucion(t, escalon(t), h)
exacta = 1 - np.exp(-t)
graficar(t, {"Convolución": y, "Solución exacta": exacta}, titulo="Salida ante escalón");
print("Error máximo de discretización:", np.max(np.abs(y - exacta)))
```

La aproximación usa una suma rectangular, por eso aparece un pequeño error, incluso en $t=0$. Al reducir el paso debe disminuir.

**Practica:** reduce a la mitad el paso temporal y compara el error máximo.
