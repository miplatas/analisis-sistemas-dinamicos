# 2.3 Solución de estados y sistemas MIMO

Para $\dot x=Ax+Bu$, $y=Cx+Du$, la solución contiene una respuesta libre debida a $x(0)$ y una respuesta forzada debida a $u(t)$. Un sistema MIMO puede tener varias entradas y salidas.

Usamos dos estados, dos entradas y dos salidas: $A=\operatorname{diag}(-1,-2)$, $B=C=I$ y $D=0$. Las entradas son constantes y las condiciones iniciales son no nulas.

**Objetivo:** separar el efecto de las entradas y del estado inicial.

```python
%matplotlib inline
import numpy as np
from scipy import signal
from sistemas_dinamicos import graficar

A = [[-1, 0], [0, -2]]
B = C = np.eye(2)
D = np.zeros((2, 2))
modelo = signal.StateSpace(A, B, C, D)
t = np.linspace(0, 5, 501)
entradas = np.column_stack((np.ones_like(t), 2*np.ones_like(t)))
_, salida, estados = signal.lsim(modelo, U=entradas, T=t, X0=[0.5, -0.5])
graficar(t, {"y1": salida[:, 0], "y2": salida[:, 1]}, titulo="Dos entradas y dos salidas");
print("Estado final aproximado:", estados[-1])
```

Ambos estados tienden a uno, pero sus transitorios difieren. El estado inicial afecta la parte libre, que desaparece para estos polos negativos.

**Practica:** repite con estado inicial cero y compara las curvas. Después añade un acoplamiento fuera de la diagonal de $A$.
