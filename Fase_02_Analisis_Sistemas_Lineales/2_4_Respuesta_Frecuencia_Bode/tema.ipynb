# 2.4 Respuesta en frecuencia y diagramas de Bode

Para una entrada sinusoidal, $G(j\omega)$ determina ganancia y desfase en régimen permanente. El diagrama de Bode muestra $20\log_{10}|G(j\omega)|$ en decibeles y la fase en grados frente a frecuencia angular $\omega$ en rad/s.

Para $G(s)=1/(s+1)$, la magnitud cae 3 dB aproximadamente en $\omega=1$ rad/s.

**Objetivo:** identificar el ancho de banda de un filtro de primer orden.

```python
%matplotlib inline
import numpy as np
from scipy import signal
from sistemas_dinamicos import graficar

modelo = signal.TransferFunction([1], [1, 1])
w = np.logspace(-2, 2, 401)
w, magnitud, fase = signal.bode(modelo, w=w)
_, ax = graficar(w, {"Magnitud": magnitud}, titulo="Bode: magnitud", xlabel="Frecuencia (rad/s)", ylabel="Magnitud (dB)")
ax.set_xscale("log")
_, ax = graficar(w, {"Fase": fase}, titulo="Bode: fase", xlabel="Frecuencia (rad/s)", ylabel="Fase (grados)")
ax.set_xscale("log")
print("Magnitud y fase a 1 rad/s:", magnitud[200], "dB,", fase[200], "grados")
```

A bajas frecuencias la ganancia se aproxima a 0 dB; a altas frecuencias disminuye aproximadamente 20 dB por década. En $1$ rad/s la fase es $-45^\circ$.
