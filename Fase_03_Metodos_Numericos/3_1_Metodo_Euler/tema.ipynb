# 3.1 Método de Euler

Para $\dot x=f(t,x)$, Euler explícito usa $x_{k+1}=x_k+h f(t_k,x_k)$. Aproxima la solución mediante la pendiente al inicio de cada paso.

En $\dot x=-x$, $x(0)=1$, la solución exacta es $e^{-t}$. El método también se aplica a modelos no lineales cambiando $f$.

**Objetivo:** visualizar el efecto de aproximar una trayectoria por pasos.

```python
%matplotlib inline
import numpy as np
from sistemas_dinamicos import graficar, integrar

t = np.linspace(0, 5, 26)
euler = integrar(lambda t, x: -x, t, [1], metodo="euler")[:, 0]
exacta = np.exp(-t)
graficar(t, {"Euler": euler, "Exacta": exacta}, titulo="Euler explícito");
print("Error máximo:", np.max(np.abs(euler - exacta)))
```

La aproximación depende del paso y se acumula a lo largo de la simulación. Para este ejemplo, Euler queda por debajo de la solución exacta.
