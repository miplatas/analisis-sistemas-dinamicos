# 3.2 Método de Runge-Kutta de cuarto orden

RK4 combina cuatro evaluaciones de la pendiente por paso: $k_1=f(t,x)$, $k_2=f(t+h/2,x+hk_1/2)$, $k_3=f(t+h/2,x+hk_2/2)$ y $k_4=f(t+h,x+hk_3)$. La actualización es $x_{k+1}=x_k+h(k_1+2k_2+2k_3+k_4)/6$.

Comparemos Euler y RK4 para $\dot x=-x$, con el mismo paso y solución exacta $e^{-t}$.

**Objetivo:** comparar precisión con igual malla.

```python
%matplotlib inline
import numpy as np
from sistemas_dinamicos import graficar, integrar

t = np.linspace(0, 5, 26)
f = lambda t, x: -x
euler = integrar(f, t, [1], metodo="euler")[:, 0]
rk4 = integrar(f, t, [1], metodo="rk4")[:, 0]
exacta = np.exp(-t)
graficar(t, {"Euler": euler, "RK4": rk4, "Exacta": exacta}, titulo="Comparación de métodos");
print("Errores máximos:", np.max(np.abs(euler-exacta)), np.max(np.abs(rk4-exacta)))
```

RK4 es más preciso en este ejemplo con igual paso, a cambio de más evaluaciones de $f$. Esto no garantiza precisión para cualquier paso o modelo.
