# Runge-Kutta de cuarto orden (RK4): péndulo amortiguado

**Objetivo:** simular el modelo desde varias condiciones iniciales e interpretar
su evolución temporal y su retrato de fase.

Con $x=(\theta,\omega)$, $\dot\theta=\omega$ y
$\dot\omega=-(g/l)\sin\theta-b\omega/(ml^2)$. Usamos $g=9.81$ m/s²,
$l=m=1$ y $b=0.6$ N·m·s/rad; el ángulo inicial principal es 2.45 rad.

RK4 combina cuatro pendientes por paso:
$k_1=f(t_k,x_k)$, $k_2=f(t_k+h/2,x_k+hk_1/2)$,
$k_3=f(t_k+h/2,x_k+hk_2/2)$ y $k_4=f(t_k+h,x_k+hk_3)$.
La actualización es $x_{k+1}=x_k+h(k_1+2k_2+2k_3+k_4)/6$.
La biblioteca implementa este método; la elección del paso debe verificarse
frente a una solución de referencia.

```python
%matplotlib inline
import numpy as np
from sistemas_dinamicos import integrar, graficar, retrato_fase

g, l, m, b = 9.81, 1.0, 1.0, 0.6
def f(t, x):
    return np.array([x[1], -g/l*np.sin(x[0])-b/(m*l**2)*x[1]])
equilibrio = (0, 0)
h = 0.005
t = np.linspace(0, 12.0, round(12.0/h) + 1)
condiciones = [[2.45, 0], [-2.45, 0], [1.2, 2.5], [-1.2, -2.5], [-np.pi+0.15, 3.8], [np.pi-0.15, -3.8]]
trayectorias = [integrar(f, t, x0, metodo="rk4") for x0 in condiciones]
principal = trayectorias[0]
print("Paso:", t[1]-t[0], "s; condición inicial:", condiciones[0])
```

## Evolución temporal

La primera condición inicial es la trayectoria principal.

```python
graficar(t, {"Ángulo θ (rad)": principal[:, 0], "Velocidad ω (rad/s)": principal[:, 1]},
         titulo="Evolución temporal: péndulo amortiguado", ylabel="Estados (ver unidades en la leyenda)");
```

## Retrato de fase

Cada punto negro marca el inicio de una trayectoria.

```python
fig, ax = retrato_fase(trayectorias, titulo="Retrato de fase: péndulo amortiguado",
                        xlabel="Ángulo θ (rad)", ylabel="Velocidad ω (rad/s)")
ax.plot(*equilibrio, "ko", label="Equilibrio de referencia")
ax.legend();
```

## Interpretación y práctica

La fricción disipa energía y el origen es localmente estable. Otros
equilibrios equivalentes ocurren cada $2\pi$; no todas las condiciones iniciales
deben acercarse al origen en un ángulo no reducido módulo $2\pi$.
