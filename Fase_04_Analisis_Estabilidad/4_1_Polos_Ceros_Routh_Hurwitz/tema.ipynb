# 4.1 Polos, ceros y criterio de Routh-Hurwitz

Los polos son raíces del denominador y los ceros raíces del numerador de una función de transferencia. En un modelo continuo LTI, los polos de una realización mínima estrictamente en el semiplano izquierdo indican estabilidad asintótica.

La tabla de Routh permite contar raíces del polinomio característico en el semiplano derecho mediante los cambios de signo en la primera columna, cuando no aparecen casos especiales.

**Objetivo:** relacionar polos y tabla de Routh para $s^3+6s^2+11s+6$.

```python
import numpy as np
from sistemas_dinamicos import routh

numerador = [1, 2]
denominador = [1, 6, 11, 6]
print("Ceros:", np.roots(numerador))
print("Polos:", np.roots(denominador))
print("Tabla de Routh:")
display(routh(denominador))
```

La primera columna es positiva y no hay cambios de signo; los polos son $-1$, $-2$ y $-3$. El cero en $-2$ cancela un polo en la relación entrada-salida; modos internos ocultos requieren analizar la realización.

La función `routh` rechaza explícitamente pivotes cero y filas nulas: esos casos necesitan el procedimiento especial de Routh y no se resuelven aquí.
