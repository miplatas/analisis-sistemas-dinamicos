# Campo de fase de un péndulo y sus linealizaciones

**Objetivo:** comparar el sistema no lineal con sus aproximaciones cerca de dos equilibrios.
El modelo con fricción viscosa es

$$\dot\theta=\omega,\qquad
\dot\omega=-\frac{g}{l}\sin\theta-\frac{b}{ml^2}\omega.$$

Usamos $g=9.81$ m/s², $l=1$ m, $m=1$ kg y $b=0.5$ N·m·s/rad.
Los equilibrios satisfacen $\omega=0$ y $\theta=k\pi$.
Alrededor de $(0,0)$, $\sin\theta\approx\theta$; alrededor de $(\pi,0)$,
$\sin\theta\approx-(\theta-\pi)$. Estas aproximaciones solo son locales.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from sistemas_dinamicos import campo_fase

g, l, m, b = 9.81, 1.0, 1.0, 0.5
amortiguamiento = b / (m*l**2)
A_0 = np.array([[0, 1], [-g/l, -amortiguamiento]])
A_pi = np.array([[0, 1], [g/l, -amortiguamiento]])
print("Autovalores en (0, 0):", np.linalg.eigvals(A_0))
print("Autovalores en (π, 0):", np.linalg.eigvals(A_pi))
```

## Campos de dirección

Las flechas tienen longitud normalizada: indican dirección, mientras que el color
muestra la magnitud numérica de las derivadas según las escalas elegidas. Como
los estados tienen unidades diferentes, esa magnitud no es una rapidez física única.
Los modelos linealizados solo tienen el equilibrio alrededor del que se construyeron.

```python
campos = [
    ("Péndulo no lineal", lambda theta, w: (w, -g/l*np.sin(theta)-amortiguamiento*w), [(-np.pi, 0), (0, 0), (np.pi, 0)]),
    ("Linealizado en (0, 0)", lambda theta, w: (w, -g/l*theta-amortiguamiento*w), [(0, 0)]),
    ("Linealizado en (π, 0)", lambda theta, w: (w, g/l*(theta-np.pi)-amortiguamiento*w), [(np.pi, 0)]),
]
for titulo, campo, equilibrios in campos:
    fig, ax = campo_fase(campo, (-1.5*np.pi, 1.5*np.pi), (-6, 6),
                         puntos=(27, 21), titulo=titulo,
                         xlabel="Ángulo θ (rad)", ylabel="Velocidad ω (rad/s)")
    for theta_eq, w_eq in equilibrios:
        ax.plot(theta_eq, w_eq, "ro")
    plt.show()
```

## Interpretación y práctica

Los autovalores en el origen tienen parte real negativa: es un foco localmente
asintóticamente estable. En $(\pi,0)$ tienen signos opuestos: es una silla
inestable. La coincidencia entre campos empeora al alejarse del equilibrio.
