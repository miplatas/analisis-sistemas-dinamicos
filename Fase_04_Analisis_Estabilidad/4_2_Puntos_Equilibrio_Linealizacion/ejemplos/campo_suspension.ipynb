# Campo de fase de una suspensión lineal

**Objetivo:** relacionar un modelo mecánico con sus autovalores y campo de fase.
Para una masa alrededor de su posición de equilibrio estático, sin excitación externa:

$$m\ddot x+c\dot x+kx=0,\quad
\dot x=v,\quad \dot v=-\frac{k}{m}x-\frac{c}{m}v.$$

Utilizamos $m=250$ kg, $k=15000$ N/m y $c=1200$ N·s/m.
El equilibrio es $(x,v)=(0,0)$.

```python
%matplotlib inline
import numpy as np
from sistemas_dinamicos import campo_fase

m, k, c = 250.0, 15000.0, 1200.0
A = np.array([[0, 1], [-k/m, -c/m]])
print("Autovalores:", np.linalg.eigvals(A))
print("Amortiguamiento crítico:", 2*np.sqrt(k*m), "N·s/m")
fig, ax = campo_fase(lambda x, v: (v, -k/m*x-c/m*v),
                     (-0.4, 0.4), (-3, 3), titulo="Campo de la suspensión",
                     xlabel="Desplazamiento x (m)", ylabel="Velocidad v (m/s)")
ax.plot(0, 0, "go", label="Equilibrio")
ax.legend();
```

## Interpretación y práctica

El amortiguamiento es inferior al crítico y los autovalores complejos tienen parte
real negativa: el origen es un foco asintóticamente estable. Las flechas muestran
dirección, no trayectorias completas. Su color combina componentes con distintas
unidades y depende de la escala elegida.

**Practica:** cambia $c$ al valor crítico y luego a uno mayor; compara autovalores
y campos. No confundas el campo de dirección con la velocidad física.
