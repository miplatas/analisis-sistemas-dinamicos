# Campo proyectado y trayectoria del sistema de Lorenz

**Objetivo:** interpretar un corte de un sistema tridimensional sin confundirlo
con un retrato de fase bidimensional autónomo.

$$\dot x=\sigma(y-x),\quad \dot y=x(\rho-z)-y,\quad
\dot z=xy-\beta z.$$

Con $\sigma=10$, $\rho=28$ y $\beta=8/3$, mostramos las componentes
$(\dot x,\dot z)$ evaluadas en el corte fijo $y=10$. El color representa la
magnitud de esas dos componentes proyectadas, no la del campo tridimensional.

```python
%matplotlib inline
import numpy as np
from scipy.integrate import solve_ivp
from sistemas_dinamicos import campo_fase

sigma, rho, beta = 10.0, 28.0, 8/3
y_corte = 10.0
def lorenz(t, estado):
    x, y, z = estado
    return [sigma*(y-x), x*(rho-z)-y, x*y-beta*z]

equilibrio = np.sqrt(beta*(rho-1))
print("Equilibrios 3D: (0, 0, 0) y (±a, ±a, ρ-1), a =", equilibrio)
```

```python
fig, ax = campo_fase(
    lambda x, z: (sigma*(y_corte-x), x*y_corte-beta*z),
    (-25, 25), (0, 50), titulo="Lorenz: corte y=10 y proyección de trayectoria",
    xlabel="Estado x", ylabel="Estado z")
sol = solve_ivp(lorenz, (0, 35), [1, 1, 1], t_eval=np.linspace(0, 35, 4000),
                rtol=1e-8, atol=1e-10)
if not sol.success:
    raise RuntimeError(sol.message)
ax.plot(sol.y[0], sol.y[2], color="black", alpha=0.35, lw=0.8,
        label="Proyección de trayectoria 3D")
ax.plot([0, -equilibrio, equilibrio], [0, rho-1, rho-1], "ro",
        label="Proyecciones de equilibrios 3D")
ax.legend();
```

## Interpretación y práctica

La curva negra es la proyección de una trayectoria tridimensional cuyo $y$
varía; **no** tiene por qué seguir las flechas del corte $y=10$. Los puntos rojos
también son proyecciones y no son equilibrios del campo del corte: ninguno tiene
$y=10$. La geometría ilustra el atractor, no demuestra caos por sí sola.

**Practica:** cambia $y_{\rm corte}$ a $-10$ y compara las flechas, manteniendo
la trayectoria tridimensional sin cambios.
