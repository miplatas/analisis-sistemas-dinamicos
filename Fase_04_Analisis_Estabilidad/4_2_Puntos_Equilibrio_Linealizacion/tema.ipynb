# 4.2 Equilibrios y linealización

Un equilibrio $x^*$ de $\dot x=f(x)$ satisface $f(x^*)=0$. 

Cerca de él, $\delta\dot x\approx J(x^*)\delta x$, donde $J=\partial f/\partial x$.

**Ejemplo:**

Considera $\dot x_1=x_2$, $\dot x_2=x_1-x_1^3-x_2$. Los autovalores del jacobiano clasifican el comportamiento local cuando ninguno tiene parte real cero.

**Objetivo:** encontrar equilibrios y estudiar su estabilidad local.

```python
%matplotlib inline
import sympy as sp
from sistemas_dinamicos import integrar, graficar
import numpy as np

x1, x2 = sp.symbols("x1 x2", real=True)
f = sp.Matrix([x2, x1 - x1**3 - x2])
J = f.jacobian([x1, x2])
equilibrios = sp.solve(list(f), [x1, x2], dict=True)
for equilibrio in equilibrios:
    print("Equilibrio:", equilibrio, "; autovalores:", J.subs(equilibrio).eigenvals())
t = np.linspace(0, 10, 501)
trayectoria = integrar(lambda t, x: np.array([x[1], x[0] - x[0]**3 - x[1]]), t, [0.8, 0])
graficar(t, {"x1": trayectoria[:, 0], "x2": trayectoria[:, 1]}, titulo="Trayectoria no lineal");
```

El origen es una silla inestable. Los equilibrios $(\pm1,0)$ tienen autovalores con parte real negativa y son localmente asintóticamente estables. La trayectoria mostrada parte cerca de $(1,0)$.

**Practica:** cambia la condición inicial y observa a qué equilibrio se aproxima. La gráfica no demuestra estabilidad global.
